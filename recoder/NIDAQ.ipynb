```python
import threading
from pathlib import Path

import lib_ipmu_daq_api as api
import lib_ipmu_daq_process as post
import lib_ipmu_daq_generator as gen
#import lib_ipmu_daq_aquisition as acq


import time

%load_ext autoreload
%autoreload 2
```

```python
app = api.DAQApp(DEBUG=True)
app.setup() # initialize widgets
app.initStorer(runs_dir=Path("./")) # initialize HDF5 storage
app.initLogger()  # Initialize logger
```

```python
generator = gen.Generator(buf_q=app.buf_q, stop_event=app.stop_event)
#acquisition = acq.DataAquisition(buf_q=app.buf_q, stop_event=app.stop_event)
processor = post.Processor(buf_q=app.buf_q, quad_q=app.quad_q, comvel_q=app.comvel_q, h5f=app.h5f, dset=app.dset, stop_event=app.stop_event, logger=app.logger, debug=app.DEBUG)
```

```python
try:
    generator_thread = threading.Thread(target=generator.run, daemon=True)
    #acquisition_thread = threading.Thread(target=acquisition.run, daemon=True)
    processor_thread = threading.Thread(target=processor.run, daemon=True)
    #acquisition_thread.start()
    generator_thread.start()
    processor_thread.start()

    time.sleep(10)
    app.stop_event.set()

finally:
    app.shutdown()
    #acquisition_thread.join(timeout=0.1)
    generator_thread.join(timeout=0.1)
    processor_thread.join(timeout=0.1)
```
